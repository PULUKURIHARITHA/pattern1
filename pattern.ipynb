```python
n=int(input("enter the size"))
for i in range(n):
    for j in range(i):
        print(j, end=' ')
    print()
```

```python
n=int(input("enter the size"))
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for k in range(2*i-1):
        print("*",end=" ")
    print()
```
